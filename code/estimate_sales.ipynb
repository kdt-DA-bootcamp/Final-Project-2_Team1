## 면적 당 매출
1. 매장의 면적 구해야 함 (국토교통부 면적 데이터)
2. 매장 각각의 면적을 구하기 어려움

- 카페의 매출을 설명하는 면적 = 카페가 그 건물에서 커버하는 영역 정도로 봄
- 즉 해당 건물에 카페가 하나 => 커버면적 = 건물의 연면적 or 건축물 면적
- 해당 건물에 카페가 n개 => 커버면적 = 건물의 연면적/n or 건축물 면적/n 

````python
import pandas as pd

# 1. 필요한 칼럼만 추출
df_sub = df[['상가업소번호', '상호명', '지점명','TRDAR_CD', '법정동코드', '지번본번지', '지번부번지', '건물관리번호']].copy()
geo_sub = geo[['고유번호', '법정동코드', '지번', '연면적', '건축물면적(㎡)']].copy()

# 2. 법정동코드 타입 맞추기 (문자열로)
df_sub['법정동코드'] = df_sub['법정동코드'].astype(str)
geo_sub['법정동코드'] = geo_sub['법정동코드'].astype(str)

# 3. 지번본번지, 지번부번지 결합해서 '지번' 만들기
df_sub['지번본번지'] = df_sub['지번본번지'].fillna(0).astype(int)
df_sub['지번부번지'] = df_sub['지번부번지'].fillna(0).astype(int)

df_sub['지번'] = df_sub.apply(
    lambda x: f"{x['지번본번지']}-{x['지번부번지']}" if x['지번부번지'] != 0 else str(x['지번본번지']),
    axis=1
)

# 4. geo_sub에서 (법정동코드 + 지번) 기준으로 건축물면적 가장 큰 것만 남김 (중복값이 있어서)
geo_sub = geo_sub.sort_values('건축물면적(㎡)', ascending=False)
geo_sub = geo_sub.drop_duplicates(subset=['법정동코드', '지번'], keep='first')

# 5. 법정동코드 + 지번 기준으로 merge
merged_df = pd.merge(
    df_sub,
    geo_sub,
    on=['법정동코드', '지번'],
    how='left'
) 

# 6. 결과를 CSV로 저장
merged_df.to_csv('temp_카페_면적_예상.csv', index=False, encoding='utf-8-sig')

# 7. 결과 확인
print(merged_df.head())
````

# 매출 추정 모델링

### 🔧 1. 모델링 목적
상권별 추정 매출을 매장 단위로 분배하거나, 매장별 상대적 매출 추정값을 예측하는 회귀 문제로 설정합니다.

### 🏗️ 2. 사용 Feature 정리 및 설명
| Feature | 설명 | 수집/처리 방식 |
|--------|------|----------------|
| **상권별 유동인구** | 상권의 잠재 고객 수 | 공공 데이터 기반 (생활인구), 매장별로 상권 내 분포 가중치 필요 |
| **매장 면적** | 매장 크기 (면적 클수록 매출 기대치 증가) | 크롤링 (부동산/임대 정보), 없으면 업종 평균 또는 상권 평균 면적 대입 |
| **프랜차이즈 여부** | 브랜드 인지도/운영력 반영 | 프랜차이즈 목록 기반 binary feature 또는 브랜드별 one-hot |
| **리뷰 수 (네이버)** | 소비자 관심/방문 수 지표 | 네이버 리뷰 수 크롤링 (카카오맵 등 병행 가능), 결측치 처리 방안 필요 |
| **경쟁도 (밀집도)** | 동일 업종 밀집도 (반경 100, 300, 500m 당 점포 수 등) | 자체 상가 데이터로 계산, 반경 내 동일 업종 수로 파생 변수 생성 |

매장별_추정매출 = 상권_총매출 * (매장별_가중치 / 가중치 총합)

### 📊 5. 추가 제안
경쟁도는 반경 200m/500m/1km 별로 만들어서 비교

리뷰 수는 로그 변환(log1p) 추천 (편향된 분포 보정)

면적이 결측일 경우 평균값 또는 비슷한 상권 평균값 대체

**LightGBM 사용으로 시작**

````python
# 리뷰 수 (review_count)
# 프랜차이즈 여부 (is_chain)
# 매장 면적 (area)
# 경쟁도 (density)
# 상권ID → 상권 단위 그룹핑용
# chatgpt 예시

import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
import lightgbm as lgb
import matplotlib.pyplot as plt

# 1. 데이터 불러오기
stores = pd.read_csv('store_info.csv')  # store_id, 상권ID, review_count, is_chain, area, density
regions = pd.read_csv('region_sales.csv')  # 상권ID, total_sales

# 2. 병합
df = stores.merge(regions, on='상권ID', how='left')

# 3. 전처리
df['log_review'] = np.log1p(df['review_count'])  # 로그 변환
df['is_chain'] = df['is_chain'].map({'Y': 1, 'N': 0})  # binary
df['inv_density'] = 1 / (df['density'] + 1)  # 경쟁이 적을수록 유리하다고 가정
df['area_missing'] = df['area'].isna().astype(int)
df['area'] = df['area'].fillna(df['area'].mean())  # 평균 대체 (또는 프랜차이즈별 평균)

# 4. 가중치 계산 (비모델 방식 → 타깃 생성용)
df['weight'] = (
    df['log_review'] * 1.0 +
    df['is_chain'] * 0.5 +
    df['area'] * 0.001 +
    df['inv_density'] * 1.0
)

# 5. 매장별 매출 타깃 생성
df['sum_weight_by_region'] = df.groupby('상권ID')['weight'].transform('sum')
df['weight_ratio'] = df['weight'] / df['sum_weight_by_region']
df['target_sales'] = df['weight_ratio'] * df['total_sales']

# 6. 모델 학습용 데이터 준비
features = ['log_review', 'is_chain', 'area', 'inv_density', 'area_missing']
target = 'target_sales'

X_train, X_test, y_train, y_test = train_test_split(df[features], df[target], test_size=0.2, random_state=42)

# 7. LightGBM 회귀 모델 학습
model = lgb.LGBMRegressor(random_state=42)
model.fit(X_train, y_train)

# 8. 예측 및 결과 저장
df['predicted_sales'] = model.predict(df[features])

# 9. 중요도 시각화
lgb.plot_importance(model, importance_type='gain', max_num_features=10)
plt.title('Feature Importance')
plt.show()

# 10. 결과 출력
df[['store_id', 'target_sales', 'predicted_sales']].head()
````

````python
import pandas as pd
from fuzzywuzzy import process

def preprocess_data(df_cafe, df_info):
    # 도로명주소 전처리
    df_info['도로명주소_전처리'] = (
        df_info['도로명주소']
        .str.split(',').str[0]
        .str.replace(r"\(.*\)", "", regex=True)
        .str.strip()
    )

    # 지점명 결측치 처리 및 전체상호명 생성
    df_cafe['지점명'] = df_cafe['지점명'].fillna("")
    df_cafe['전체상호명'] = df_cafe['상호명'] + df_cafe['지점명']

    return df_cafe, df_info

def fuzzy_match_merge(df_cafe, df_info, threshold=70):
    match_list = []

    for idx, addr in df_cafe['도로명주소'].items():
        query = df_cafe.loc[idx, '전체상호명']
        match_info = df_info[df_info['도로명주소_전처리'] == addr]

        if not match_info.empty:
            choices = match_info['사업장명'].tolist()
            best_match = process.extractOne(query, choices)

            if best_match:
                best, score = best_match
                if score >= threshold:
                    matched_row = match_info[match_info['사업장명'] == best].iloc[0]
                    match_list.append({
                        'index': idx,
                        'matched_name': best,
                        'score': score,
                        '시설총규모': matched_row['시설총규모'],
                        '사업장명': matched_row['사업장명'],
                        '보증액': matched_row['보증액'],
                        '월세액': matched_row['월세액']
                    })
                    print(f'{best, df_cafe.loc[idx,'전체상호명']} 추가 완료!')
            else:
                print(f'{query}추가 실패')

    # 결과 병합
    match_df = pd.DataFrame(match_list).set_index('index')
    df_merged = df_cafe.merge(match_df, left_index=True, right_index=True, how='left')

    return df_merged

# 🔁 사용 예시
# 1. 데이터 전처리
df_cafe, df_info = preprocess_data(df_cafe, df_info)

# 2. 병합 실행 (threshold=70)
df_result = fuzzy_match_merge(df_cafe, df_info, threshold=50)

# 3. 결과 확인
print(df_result.shape)
df_result[['전체상호명', '사업장명', 'score', '시설총규모', '보증액', '월세액']].head()
print(df_result.head(3))
````
