```python
import pandas as pd

df_mapping = pd.read_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\data\데이터정제\카페_상권_매핑_데이터_중복제거.csv')
df_mapping.rename(columns={'TRDAR_CD':'상권_코드', 'TRDAR_CD_N':'상권_코드_명'}, inplace=True)
# 접근성 피처 (통합)
df_all = pd.read_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\data\매물추천\카페_통합_최종데이터.csv') 
# 인구 피처
df_pop = pd.read_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\data\매물추천\feature_population_0505.csv') # 인구 결측치 (면적 제공 x?)
df_day = pd.read_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\data\매물추천\평일_주말_유동인구_차이.csv')
# 성공 여부 결합
df_sucess = pd.read_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\data\점수산출\카페_성공여부_다중.csv')
df_sucess.rename(columns={'sucess':'성공_여부'}, inplace=True)
# 시인성
df_vis = pd.read_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\data\데이터정제\카페_시인성정보.csv')
# 집객시설 
df_ven = pd.read_csv(r"C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\data\대용량\서울시 상권분석서비스(집객시설-상권).csv", encoding = 'euc-kr')
df_ven['기준_년분기_코드'] = df_ven['기준_년분기_코드'].astype(str)
df_ven = df_ven[df_ven['기준_년분기_코드'] == '20244']

map_cols = ['상가업소번호','상호명','지점명','상권_코드','상권_코드_명','시군구코드','시군구명','행정동코드','행정동명','법정동코드','법정동명','지번코드','지번본번지','지번부번지','지번주소','도로명코드','도로명','건물본번지','건물부번지','도로명주소','신우편번호','층정보',"호정보",'경도','위도']
all_cols = ['상가업소번호','상권_코드','가장가까운_지하철역', '지하철역_거리(m)', '가장가까운_버스정류장', '버스정류장_거리(m)', '반경500m내_카페수', '밀집등급']
pop_cols = ['상가업소번호','상권_코드','est_총_유동인구_수','피크_유동인구_수_상권별', '피크_유동인구_수_11_14', '매출_최대_시간대','est_시간대_00_06_유동인구_수','est_시간대_06_11_유동인구_수','est_시간대_11_14_유동인구_수','est_시간대_14_17_유동인구_수','est_시간대_17_21_유동인구_수','est_시간대_21_24_유동인구_수']
day_cols = ['상권_코드', '평균_평일_유동인구_수','평균_주말_유동인구_수','평일_대비_주말_유동인구_비율']
suc_cols = ['상가업소번호','총점','성공_여부']
vis_cols = ['상가업소번호','접한길이_m','도로2m이내접촉길이_m','교차로_존재_여부']
ven_cols = ['상권_코드',"집객시설_수","관공서_수","은행_수","종합병원_수","일반_병원_수","약국_수","유치원_수","초등학교_수","중학교_수","고등학교_수","대학교_수","백화점_수","슈퍼마켓_수","극장_수","숙박_시설_수","지하철_역_수","버스_정거장_수"]

df_merged = pd.merge(df_mapping[map_cols], df_all[all_cols], on = ['상가업소번호', '상권_코드'], how = 'left') # 중복없이 merge 완
df_merged = pd.merge(df_merged, df_pop[pop_cols], how = 'left', on = ['상가업소번호', '상권_코드']) 
df_merged = df_merged.merge(df_day[day_cols], how = 'left', on = '상권_코드')
df_merged = df_merged.merge(df_sucess[suc_cols], how = 'left', on = '상가업소번호')
df_merged = df_merged.merge(df_vis[vis_cols], how = 'left', on = '상가업소번호')
df_merged = df_merged.merge(df_ven[ven_cols], how = 'left', on = '상권_코드')
df_merged.columns.tolist()
df_merged.to_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\data\매물추천\카페_매물추천피처_다중.csv', encoding = 'utf-8-sig', index = False)
```
