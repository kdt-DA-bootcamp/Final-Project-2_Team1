```python
import pandas as pd
import requests
import os
import time
from dotenv import load_dotenv
from rapidfuzz import fuzz

# 📂 환경 변수 로드
load_dotenv()
KAKAO_API_KEY = os.getenv("KAKAO_REST_API_KEY")

# 📂 데이터 로드
df = pd.read_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\code\카카오ID 부착 중.csv')

# ❗ NaN인 행만 필터링
df_missing = df[df['카카오맵ID'].isna()].copy()

# 📍 위경도 + 상호명 기반으로 카카오맵 ID 재시도
def get_place_id_by_coords(lat, lon, name):
    url = 'https://dapi.kakao.com/v2/local/search/category.json'
    headers = {
        "Authorization": f"KakaoAK {KAKAO_API_KEY}"
    }
    params = {
        "category_group_code": "CE7",  # 카페
        "x": lon,
        "y": lat,
        "radius": 100,
        "size": 5
    }

    try:
        response = requests.get(url, headers=headers, params=params)
        if response.status_code != 200:
            print(f"❌ API 오류: {response.status_code}")
            return None

        data = response.json()
        candidates = data.get('documents', [])

        for place in candidates:
            similarity = fuzz.partial_ratio(name, place['place_name'])
            if similarity >= 80:
                return place['id']
        return None
    except Exception as e:
        print(f"⚠️ 예외 발생: {e}")
        return None

# 🔁 반복 수행
print(f"🔍 보완 시도 중... 총 {len(df_missing)}건")

for i in df_missing.index:
    lat = df_missing.loc[i, '위도']
    lon = df_missing.loc[i, '경도']
    name = df_missing.loc[i, '상호명']
    
    place_id = get_place_id_by_coords(lat, lon, name)
    df_missing.loc[i, '카카오맵ID'] = place_id

    if i % 10 == 0:
        print(f"🔄 {i}번째까지 진행됨")
    time.sleep(0.3)

# 🧩 원본 df에 반영
df.update(df_missing)

# 💾 저장
output_path = r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\code\카카오ID_보완_적용됨.csv'
df.to_csv(output_path, index=False, encoding='utf-8-sig')
print(f"✅ 저장 완료: {output_path}")
```

```python
import pandas as pd
import requests
import os
from dotenv import load_dotenv
from rapidfuzz import fuzz

# 환경 변수 불러오기
load_dotenv()
KAKAO_API_KEY = os.getenv("KAKAO_REST_API_KEY")

# 카카오 API 함수: 도로명주소 기반 ID 찾기 + 상호명 유사도 비교
def get_place_id_by_address(address, name, threshold=70):
    url = 'https://dapi.kakao.com/v2/local/search/address.json'
    headers = {
        "Authorization": f"KakaoAK {KAKAO_API_KEY}"
    }
    params = {
        "query": address
    }

    try:
        response = requests.get(url, headers=headers, params=params)
        response.raise_for_status()
        data = response.json()

        if not data.get('documents'):
            return None

        lat = data['documents'][0]['y']
        lng = data['documents'][0]['x']

        keyword_url = 'https://dapi.kakao.com/v2/local/search/keyword.json'
        keyword_params = {
            "query": name,
            "x": lng,
            "y": lat,
            "radius": 100,
            "size": 10
        }
        keyword_resp = requests.get(keyword_url, headers=headers, params=keyword_params)
        keyword_resp.raise_for_status()
        keyword_data = keyword_resp.json()

        for place in keyword_data.get('documents', []):
            similarity = fuzz.partial_ratio(name, place['place_name'])
            if similarity >= threshold:
                return place['id']
        return None

    except Exception as e:
        print(f"❌ 오류 발생: {e}")
        return None

# 파일 불러오기
input_file = "카카오ID_보완_적용됨_수정버전.csv"
df = pd.read_csv(input_file)

# 보완 시도
updated_count = 0
for i in range(len(df)):
    if pd.isna(df.loc[i, '카카오맵ID']):
        address = df.loc[i, '도로명주소']
        name = df.loc[i, '상호명']
        kakao_id = get_place_id_by_address(address, name)
        if kakao_id:
            df.loc[i, '카카오맵ID'] = kakao_id
            updated_count += 1

print(f"📌 유사한 상호명을 기준으로 보완된 ID 수: {updated_count}개")

# 결과 저장
final_path = "카카오ID_보완_적용됨_최종_수정버전.csv"
df.to_csv(final_path, index=False, encoding='utf-8-sig')

# 실패 로그 저장
failed_rows = df[df['카카오맵ID'].isna()]
failed_rows.to_csv("카카오ID_보완_최종_실패_로그.csv", index=False, encoding='utf-8-sig')

print("✅ 도로명주소 기반 유사도 보완 완료! 최종 파일과 실패 로그 저장됨.")
```

```python
import pandas as pd
import requests
import time
import os
import re
from difflib import SequenceMatcher
from dotenv import load_dotenv

load_dotenv()
KAKAO_REST_API_KEY = os.getenv("KAKAO_REST_API_KEY")

# 문자열 유사도 계산
def get_similarity(a, b):
    return SequenceMatcher(None, a.strip(), b.strip()).ratio()

# 도로명주소에서 '서울특별시 구' 추출
def extract_seoul_gu(address):
    match = re.search(r"(서울특별시\s+\S+구)", address)
    return match.group(1).strip() if match else None

# 검색 결과 주소에서 '서울 구' 추출
def extract_kakao_gu(address):
    match = re.search(r"(서울\s+\S+구)", address)
    return match.group(1).strip().replace(" ", "") if match else None

# 카카오 검색 함수
def search_kakao(query, target_name, target_gu):
    if not query or query.lower() == "nan":
        return None

    url = "https://dapi.kakao.com/v2/local/search/keyword.json"
    headers = {"Authorization": f"KakaoAK {KAKAO_REST_API_KEY}"}
    params = {"query": query}

    try:
        response = requests.get(url, headers=headers, params=params)
        if response.status_code == 200:
            places = response.json().get("documents", [])
            for place in places:
                addr = place.get("address_name", "")
                if "서울" not in addr:
                    continue

                kakao_gu = extract_kakao_gu(addr)
                if not kakao_gu or not target_gu:
                    continue

                if kakao_gu != target_gu.replace(" ", ""):
                    continue

                similarity = get_similarity(target_name, place["place_name"])
                if similarity >= 0.5:
                    return place["id"]
    except Exception as e:
        print(f"[API 오류] {query} - {e}")
    return None

# 데이터 불러오기
file_path = r"C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\code\카카오ID_보완_적용됨_최종_수정버전_지번주소.csv"
df = pd.read_csv(file_path, encoding="utf-8-sig")
df_nan = df[df["카카오맵ID"].isna()].copy()
fail_log = []

# 검색 및 보완
for idx in df_nan.index:
    row = df.loc[idx]
    name = str(row["상호명"]).strip()
    branch = str(row["지점명"]).strip() if pd.notna(row["지점명"]) else ""
    query = f"{name} {branch}".strip()
    road_addr = str(row["도로명주소"])

    gu_full = extract_seoul_gu(road_addr)
    if not gu_full:
        fail_log.append(row.to_dict())
        continue

    place_id = search_kakao(query, name, gu_full)
    if place_id:
        df.at[idx, "카카오맵ID"] = place_id
    else:
        fail_log.append(row.to_dict())

    time.sleep(0.3)

# 결과 저장
output_path = file_path.replace(".csv", "_지점명포함_서울구정확일치.csv")
fail_path = file_path.replace(".csv", "_지점명포함_서울구정확일치_fail.csv")

df.to_csv(output_path, index=False, encoding="utf-8-sig")
pd.DataFrame(fail_log).to_csv(fail_path, index=False, encoding="utf-8-sig")

print(f"📌 보완 완료. NaN 남은 수: {df['카카오맵ID'].isna().sum()}건")
print(f"✅ 저장 파일: {output_path}")
print(f"❌ 실패 파일: {fail_path}")
```

```python
import pandas as pd
from selenium import webdriver
from selenium.webdriver.common.by import By
import time
import re
from webdriver_manager.chrome import ChromeDriverManager
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.chrome.service import Service

# CSV 파일 읽기
df = pd.read_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\code\kakaoID_카페_상권_매칭.csv')  # 실제 파일 경로로 변경

# 카카오맵ID를 숫자 형식으로 처리하고 .0 제거 후 int로 변환
def clean_kakao_map_id(place_id):
    # 숫자 부분만 남기고 .0 제거
    place_id = str(place_id).replace('.0', '')  # .0이 있으면 제거
    try:
        return int(place_id)  # 정수로 변환
    except ValueError:
        return None  # 변환할 수 없으면 None 반환

# 카카오맵ID에 대해 처리
df['카카오맵ID'] = df['카카오맵ID'].apply(clean_kakao_map_id)

# 셀레니움 옵션 설정
options = Options()
options.add_argument('--no-sandbox')
options.add_argument('--disable-dev-shm-usage')
service = Service(ChromeDriverManager().install())
driver = webdriver.Chrome(service=service, options=options)

# 카카오맵 ID로 별점과 후기 수 추출 함수
def get_rating_and_reviews(place_id):
    url = f'https://place.map.kakao.com/{place_id}#comment'
    driver.get(url)
    time.sleep(5)  # 페이지가 완전히 로드될 때까지 대기

    try:
        # XPath로 평균 별점 가져오기
        rating = driver.find_element(By.XPATH, "/html/body/div[2]/main/article/div[2]/div[2]/div[2]/div[1]/div[1]/span/span[2]").text
        # XPath로 다른 정보 가져오기 (예: 'strong' 태그 안의 텍스트)
        other_info = driver.find_element(By.XPATH, "/html/body/div[2]/main/article/div[2]/div[2]/div[2]/div[1]/div[1]/a/strong").text
        # '후기 7' 형태에서 숫자만 추출
        review_count = re.findall(r'\d+', other_info)[0]  # '후기 7'에서 7만 추출
        return rating, review_count
    except Exception as e:
        print(f"❌ {place_id} 데이터 추출 실패: {e}")
        return None, None

# 각 카카오맵ID에 대해 별점과 후기 수를 추가
ratings = []
review_counts = []

# 총 10개씩 배치로 처리
batch_size = 10
for i in range(0, len(df), batch_size):  # 배치 크기만큼 반복
    batch_df = df.iloc[i:i + batch_size]  # 10개씩 데이터 추출

    for place_id in batch_df['카카오맵ID']:  # 각 카카오맵ID에 대해 처리
        if pd.notna(place_id):  # NaN이 아닌 경우에만 처리
            rating, review_count = get_rating_and_reviews(place_id)
        else:
            rating, review_count = None, None
        
        ratings.append(rating)
        review_counts.append(review_count)

    # 10개 처리 후 업데이트된 데이터를 바로 저장
    df.loc[i:i + batch_size - 1, '별점'] = ratings[-batch_size:]  # 최근 배치에 해당하는 별점 추가
    df.loc[i:i + batch_size - 1, '후기 수'] = review_counts[-batch_size:]  # 최근 배치에 해당하는 후기 수 추가

    # 저장 (중간 저장)
    df.to_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\code\kakaoID_카페_상권_매칭_with_ratings.csv', index=False, encoding='utf-8-sig')
    print(f"{i + batch_size}개의 데이터를 저장했습니다.")

# 모든 배치 처리 후 드라이버 종료
driver.quit()

print("모든 작업이 완료되었습니다.")
```
