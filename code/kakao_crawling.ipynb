```python
import pandas as pd
import requests
import os
import time
from dotenv import load_dotenv
from rapidfuzz import fuzz

# 📂 환경 변수 로드
load_dotenv()
KAKAO_API_KEY = os.getenv("KAKAO_REST_API_KEY")

# 📂 데이터 로드
df = pd.read_csv(r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\code\카카오ID 부착 중.csv')

# ❗ NaN인 행만 필터링
df_missing = df[df['카카오맵ID'].isna()].copy()

# 📍 위경도 + 상호명 기반으로 카카오맵 ID 재시도
def get_place_id_by_coords(lat, lon, name):
    url = 'https://dapi.kakao.com/v2/local/search/category.json'
    headers = {
        "Authorization": f"KakaoAK {KAKAO_API_KEY}"
    }
    params = {
        "category_group_code": "CE7",  # 카페
        "x": lon,
        "y": lat,
        "radius": 100,
        "size": 5
    }

    try:
        response = requests.get(url, headers=headers, params=params)
        if response.status_code != 200:
            print(f"❌ API 오류: {response.status_code}")
            return None

        data = response.json()
        candidates = data.get('documents', [])

        for place in candidates:
            similarity = fuzz.partial_ratio(name, place['place_name'])
            if similarity >= 80:
                return place['id']
        return None
    except Exception as e:
        print(f"⚠️ 예외 발생: {e}")
        return None

# 🔁 반복 수행
print(f"🔍 보완 시도 중... 총 {len(df_missing)}건")

for i in df_missing.index:
    lat = df_missing.loc[i, '위도']
    lon = df_missing.loc[i, '경도']
    name = df_missing.loc[i, '상호명']
    
    place_id = get_place_id_by_coords(lat, lon, name)
    df_missing.loc[i, '카카오맵ID'] = place_id

    if i % 10 == 0:
        print(f"🔄 {i}번째까지 진행됨")
    time.sleep(0.3)

# 🧩 원본 df에 반영
df.update(df_missing)

# 💾 저장
output_path = r'C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\code\카카오ID_보완_적용됨.csv'
df.to_csv(output_path, index=False, encoding='utf-8-sig')
print(f"✅ 저장 완료: {output_path}")
```

```python
import pandas as pd
import requests
import os
from dotenv import load_dotenv
from rapidfuzz import fuzz

# 환경 변수 불러오기
load_dotenv()
KAKAO_API_KEY = os.getenv("KAKAO_REST_API_KEY")

# 카카오 API 함수: 도로명주소 기반 ID 찾기 + 상호명 유사도 비교
def get_place_id_by_address(address, name, threshold=70):
    url = 'https://dapi.kakao.com/v2/local/search/address.json'
    headers = {
        "Authorization": f"KakaoAK {KAKAO_API_KEY}"
    }
    params = {
        "query": address
    }

    try:
        response = requests.get(url, headers=headers, params=params)
        response.raise_for_status()
        data = response.json()

        if not data.get('documents'):
            return None

        lat = data['documents'][0]['y']
        lng = data['documents'][0]['x']

        keyword_url = 'https://dapi.kakao.com/v2/local/search/keyword.json'
        keyword_params = {
            "query": name,
            "x": lng,
            "y": lat,
            "radius": 100,
            "size": 10
        }
        keyword_resp = requests.get(keyword_url, headers=headers, params=keyword_params)
        keyword_resp.raise_for_status()
        keyword_data = keyword_resp.json()

        for place in keyword_data.get('documents', []):
            similarity = fuzz.partial_ratio(name, place['place_name'])
            if similarity >= threshold:
                return place['id']
        return None

    except Exception as e:
        print(f"❌ 오류 발생: {e}")
        return None

# 파일 불러오기
input_file = "카카오ID_보완_적용됨_수정버전.csv"
df = pd.read_csv(input_file)

# 보완 시도
updated_count = 0
for i in range(len(df)):
    if pd.isna(df.loc[i, '카카오맵ID']):
        address = df.loc[i, '도로명주소']
        name = df.loc[i, '상호명']
        kakao_id = get_place_id_by_address(address, name)
        if kakao_id:
            df.loc[i, '카카오맵ID'] = kakao_id
            updated_count += 1

print(f"📌 유사한 상호명을 기준으로 보완된 ID 수: {updated_count}개")

# 결과 저장
final_path = "카카오ID_보완_적용됨_최종_수정버전.csv"
df.to_csv(final_path, index=False, encoding='utf-8-sig')

# 실패 로그 저장
failed_rows = df[df['카카오맵ID'].isna()]
failed_rows.to_csv("카카오ID_보완_최종_실패_로그.csv", index=False, encoding='utf-8-sig')

print("✅ 도로명주소 기반 유사도 보완 완료! 최종 파일과 실패 로그 저장됨.")
```

```python
import pandas as pd
import requests
from difflib import SequenceMatcher
import time
import os
from dotenv import load_dotenv

# .env 파일에서 환경 변수 로드
load_dotenv()
KAKAO_REST_API_KEY = os.getenv("KAKAO_REST_API_KEY")

# 함수: 유사도 계산
def get_similarity(a, b):
    return SequenceMatcher(None, a, b).ratio()

# 함수: 카카오맵 장소 검색 (좌표 기반)
def search_kakao_places_by_coords(lat, lon, keyword=None):
    url = "https://dapi.kakao.com/v2/local/search/keyword.json"
    headers = {"Authorization": f"KakaoAK {KAKAO_REST_API_KEY}"}
    params = {
        "x": lon,
        "y": lat,
        "radius": 200,
        "sort": "distance",
        "query": keyword if keyword else ""
    }
    response = requests.get(url, headers=headers, params=params)
    if response.status_code == 200:
        return response.json().get("documents", [])
    else:
        return []

# 데이터 불러오기
df = pd.read_csv(r"C:\Users\iq750\bootcamp_git\Final-Project-2_Team1\data\카페_상권_매핑_데이터_0421.csv", encoding="utf-8-sig")

# 결과 저장 컬럼
df["카카오맵ID"] = pd.NA
fail_log = []

# 반복 처리
for idx, row in df.iterrows():
    name = str(row["상호명"])
    lat = row["위도"]
    lon = row["경도"]

    try:
        results = search_kakao_places_by_coords(lat, lon, keyword=name)
        matched_id = None
        max_similarity = 0

        for place in results:
            similarity = get_similarity(name, place["place_name"])
            if similarity >= 0.5 and similarity > max_similarity:
                matched_id = place["id"]
                max_similarity = similarity

        df.at[idx, "카카오맵ID"] = matched_id if matched_id else pd.NA

        if pd.isna(matched_id):
            fail_log.append(row.to_dict())

    except Exception as e:
        print(f"오류 발생 - 인덱스 {idx}: {e}")
        fail_log.append(row.to_dict())
        continue

    time.sleep(0.3)  # API rate limit 회피용 딜레이

# 결과 저장
output_file = "output_with_kakao_id_수정버전.csv"
fail_file = "fail_log_수정버전.csv"

df.to_csv(output_file, index=False, encoding="utf-8-sig")
pd.DataFrame(fail_log).to_csv(fail_file, index=False, encoding="utf-8-sig")

# NaN 개수 출력
nan_count = df["카카오맵ID"].isna().sum()
print(f"\n❗️카카오맵ID가 매칭되지 않은 항목 개수: {nan_count}")
print(f"✅ 저장 완료: {output_file}, 실패 로그: {fail_file}")
```

```python
# 보완 1. 우편번호로 검색해 보기
import pandas as pd
import requests
import time
import os
from difflib import SequenceMatcher
from dotenv import load_dotenv

# 환경변수 로드
load_dotenv()
KAKAO_REST_API_KEY = os.getenv("KAKAO_REST_API_KEY")

# 유사도 함수
def get_similarity(a, b):
    return SequenceMatcher(None, a, b).ratio()

# 카카오맵 검색 (우편번호 + 상호명)
def search_kakao_by_postcode(name, postcode):
    if pd.isna(postcode) or not str(postcode).strip():
        return None
    
    # 4자리 우편번호는 앞에 0 붙이기
    postcode = str(postcode).zfill(5)

    query = f"{name} {postcode}"
    url = "https://dapi.kakao.com/v2/local/search/keyword.json"
    headers = {"Authorization": f"KakaoAK {KAKAO_REST_API_KEY}"}
    params = {"query": query}

    try:
        response = requests.get(url, headers=headers, params=params)
        if response.status_code == 200:
            for place in response.json().get("documents", []):
                similarity = get_similarity(name, place["place_name"])
                if similarity >= 0.5:
                    return place["id"]
    except Exception as e:
        print(f"API 오류 (우편번호 기반) - {name}: {e}")
    return None

# 데이터 불러오기
df = pd.read_csv("output_with_kakao_id_수정버전.csv", encoding="utf-8-sig")

# 로그 초기화
fail_log_postcode = []

# NaN 항목만 필터링
mask_nan = df["카카오맵ID"].isna()
df_nan = df[mask_nan].copy()

for idx in df_nan.index:
    row = df.loc[idx]
    name = str(row["상호명"])
    postcode = str(row["신우편번호"])

    matched_id = search_kakao_by_postcode(name, postcode)

    if matched_id:
        df.at[idx, "카카오맵ID"] = matched_id
    else:
        fail_log_postcode.append(row.to_dict())

    time.sleep(0.3)  # API 제한 회피

# 저장
output_file = "output_with_kakao_id_우편번호추가_수정버전.csv"
fail_file = "fail_log_우편번호기반_수정버전.csv"

df.to_csv(output_file, index=False, encoding="utf-8-sig")
pd.DataFrame(fail_log_postcode).to_csv(fail_file, index=False, encoding="utf-8-sig")

# 결과 출력
nan_count = df["카카오맵ID"].isna().sum()
print(f"\n📌 남은 NaN 개수: {nan_count}")
print(f"✅ 저장 완료: {output_file}, 실패 로그: {fail_file}")
```
